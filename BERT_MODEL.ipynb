# Step 1: Install the required library
# Run this in terminal or notebook once
# pip install transformers torch

```python
from transformers import AutoTokenizer, AutoModel
import torch
```

Step 2: Load a small BERT model and its tokenizer
# Tokenizer = breaks text into small pieces (tokens) BERT understands
# Model = the brain that creates embeddings

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased")
```

# Step 3: Write a sentence


```python
sentence = "Artificial Intelligence is transforming the world."
```

# Step 4: Convert the sentence into tokens BERT understands


```python
inputs = tokenizer(sentence, return_tensors="pt")
```

# Step 5: Pass tokens to BERT to get embeddings

```python
with torch.no_grad(): # no training, just getting results
 outputs = model(**inputs)
```

# Step 6: Get the embedding vector for the whole sentence
# outputs.last_hidden_state = embeddings for each word
# mean(1) = average all word embeddings to get sentence embedding

```python
sentence_embedding = outputs.last_hidden_state.mean(dim=1)
```

# Step 7: Show the embedding vector

```python
print("Sentence Embedding Vector:\n", sentence_embedding)
print("\nVector Shape:", sentence_embedding.shape)
```
